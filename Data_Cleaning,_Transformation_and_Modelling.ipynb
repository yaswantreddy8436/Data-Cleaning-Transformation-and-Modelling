1 Data Loading is the process of connecting to an external source and importing its information into Power BI. It is considered the critical first step because you cannot clean, model, or visualize insights without first bringing the raw records into the system

2 Load Imports records directly as they currently exist. Skips the initial filtering and restructuring phase.Can result in bloated, slower models if raw files are messy.
Transform Data Opens the Power Query Editor first. Allows removing columns, fixing types, and replacing nulls.

3 A Fact Table stores quantitative measurements, while a Dimension Table stores attributes that filter and describe those measurements.
Examples from your CSV: Confirmed_Cases, Recovered_Cases, Deaths, Active_Cases, Tests_Conducted, and Hospitalized

4 Star Schema minimizes complex table joins by connecting dimension tables directly to a central fact table. Snowflake schema splits dimensions into sub-tables, which requires Power BI to process multiple nested relationships. Star schema reduces memory usage, speeds up DAX calculations, and makes it much easier to build clean visuals without confusing relationships.
